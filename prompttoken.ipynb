## Étape 3: Tokenisation avec Tiktoken

Testons la tokenisation d'un prompt avec le tokenizer OpenAI.

```python
import tiktoken

# Obtenir le tokenizer de GPT-4o
try:
    tokenizer = tiktoken.encoding_for_model("gpt-4o")
    print(f"Tokenizer trouvé: {tokenizer.name}")
except Exception as e:
    print(f"Utilisation du tokenizer o200k_base par défaut: {e}")
    tokenizer = tiktoken.get_encoding("o200k_base")

# Notre prompt de test
prompt = "Vous êtes un expert en marketing digital. C'est quoi un digital marketer ?"

print(f"\n📝 Prompt original:\n{prompt}")
print(f"\n📊 Longueur du prompt: {len(prompt)} caractères")

# Encoder le prompt
tokens = tokenizer.encode(prompt)
print(f"\n🔢 Tokens: {tokens}")
print(f"\n📈 Nombre total de tokens: {len(tokens)}")

# Afficher chaque token décodé
print(f"\n🔤 Décodage token par token:")
decoded_tokens = []
for token in tokens:
    t = tokenizer.decode_single_token_bytes(token).decode("utf-8")
    decoded_tokens.append(t)
    print(t, end='|')

print(f"\n\n📊 Ratio tokens/caractères: {len(tokens)}/{len(prompt)} = {len(tokens)/len(prompt):.3f}")
```

## Étape 5: Test OpenAI (GPT-4o)

Testons l'API OpenAI pour générer une réponse sur le marketing digital.

```python
from langchain_openai import ChatOpenAI
from IPython.display import Markdown, display

try:
    # Initialiser le LLM OpenAI
    llm = ChatOpenAI(model="gpt-4o", temperature=0)
    
    # Notre prompt
    system_message = "Vous êtes un expert en marketing digital"
    user_message = "C'est quoi un digital marketer ?"
    
    # Appeler le LLM
    resp = llm.invoke(input=[
        {"role": "system", "content": system_message},
        {"role": "user", "content": user_message}
    ])
    
    print("="*80)
    print("🤖 RÉPONSE DE GPT-4o")
    print("="*80)
    print(resp.content)
    print("\n" + "="*80)
    
    # Affichage formaté avec Markdown
    display(Markdown("### 📄 Affichage formaté:\n\n" + resp.content))
    
except Exception as e:
    print(f"❌ Erreur OpenAI: {e}")
    print("\n💡 Assurez-vous que:")
    print("1. Votre clé API est valide")
    print("2. Vous avez des crédits sur votre compte OpenAI")
    print("3. Votre clé est bien configurée dans les secrets Colab")
```

## Étape 6: Test Ollama (Local)

⚠️ **NOTE:** Ollama fonctionne localement sur votre ordinateur.

Pour tester Ollama:
1. Installez Ollama depuis https://ollama.ai
2. Dans votre terminal local, lancez: `ollama run qwen3.5:cloud`
3. Assurez-vous que Ollama est accessible sur http://localhost:11434
4. Puis exécutez la cellule ci-dessous

```python
from langchain_ollama import ChatOllama

try:
    # Initialiser le LLM Ollama
    print("⏳ Connexion à Ollama sur http://localhost:11434...")
    llm2 = ChatOllama(model="qwen3.5:cloud", temperature=0)
    
    # Notre prompt (identique à OpenAI)
    system_message = "Vous êtes un expert en marketing digital"
    user_message = "C'est quoi un digital marketer ?"
    
    # Appeler le LLM
    print("🔄 Traitement en cours...")
    resp2 = llm2.invoke(input=[
        {"role": "system", "content": system_message},
        {"role": "user", "content": user_message}
    ])
    
    print("="*80)
    print("🤖 RÉPONSE D'OLLAMA (qwen3.5)")
    print("="*80)
    print(resp2.content)
    print("\n" + "="*80)
    
    # Affichage formaté
    display(Markdown("### 📄 Affichage formaté:\n\n" + resp2.content))
    
except Exception as e:
    print(f"❌ Erreur Ollama: {e}")
    print("\n💡 Solutions:")
    print("1. Assurez-vous qu'Ollama est installé et lancé localement")
    print("2. Vérifiez que Ollama est accessible sur http://localhost:11434")
    print("3. Lancez: ollama run qwen3.5:cloud")
    print("\n⚠️ NOTE: Ollama ne fonctionne pas sur Google Colab directement")
    print("Vous devez tester cela sur votre ordinateur local")
```

## Étape 8: Tests Avancés - Sentiment Analysis

Testons l'analyse de sentiment avec nos LLM.

```python
# Définir un système de classification de sentiments
sentiment_system = """Vous êtes un expert en analyse de sentiment.
Pour chaque texte donné, vous devez:
1. Déterminer le sentiment (Positif, Négatif, Neutre)
2. Donner un score de confiance (0-100)
3. Justifier votre analyse en 1-2 phrases

Répondez au format:
Sentiment: [Positif/Négatif/Neutre]
Confiance: [score]%
Justification: [explication]"""

# Textes de test
test_texts = [
    "J'adore ce produit! C'est incroyable et je le recommande vivement à tous.",
    "Ce service est horrible. J'ai eu une mauvaise expérience et je suis très déçu.",
    "Le produit a été livré hier. Il fonctionne correctement."
]

print("="*80)
print("🎯 TEST: ANALYSE DE SENTIMENT")
print("="*80)

try:
    for i, text in enumerate(test_texts, 1):
        print(f"\n📝 Texte {i}: {text}")
        print("-" * 80)
        
        resp = llm.invoke(input=[
            {"role": "system", "content": sentiment_system},
            {"role": "user", "content": text}
        ])
        
        print("🤖 GPT-4o:")
        print(resp.content)
        
except Exception as e:
    print(f"❌ Erreur: {e}")
```
