## Etape 3: Tokenisation avec Tiktoken

Testons la tokenisation d'un prompt pour comprendre comment les LLMs lisent le texte.

```python
import tiktoken

# Obtenir le tokenizer
try:
    tokenizer = tiktoken.encoding_for_model("gpt-4o")
    print(f"Tokenizer trouve: {tokenizer.name}")
except Exception as e:
    print(f"Utilisation du tokenizer o200k_base par defaut: {e}")
    tokenizer = tiktoken.get_encoding("o200k_base")

prompt = "Vous etes un expert en marketing digital. C'est quoi un digital marketer ?"

print(f"\nPrompt original:\n{prompt}")
print(f"\nLongueur du prompt: {len(prompt)} caracteres")

tokens = tokenizer.encode(prompt)
print(f"\nTokens: {tokens}")
print(f"\nNombre total de tokens: {len(tokens)}")

print(f"\nDecodage token par token:")
decoded_tokens = []
for token in tokens:
    t = tokenizer.decode_single_token_bytes(token).decode("utf-8")
    decoded_tokens.append(t)
    print(t, end='|')

print(f"\n\nRatio tokens/caracteres: {len(tokens)}/{len(prompt)} = {len(tokens)/len(prompt):.3f}")
```

## Etape 5: Test Gemini

Testons l'API Gemini pour generer une reponse sur le marketing digital.

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from IPython.display import Markdown, display

try:
    # Initialiser le LLM Gemini
    llm = ChatGoogleGenerativeAI(model="gemini-2.0-flash", temperature=0)

    # Notre prompt format tuple pour LangChain
    messages = [
        ("system", "Vous etes un expert en marketing digital"),
        ("human", "C'est quoi un digital marketer ?")
    ]

    # Appeler le LLM
    print("Generation de la reponse en cours...\n")
    resp = llm.invoke(messages)

    print("="*80)
    print("REPONSE DE GEMINI")
    print("="*80)

    # Affichage formate avec Markdown
    display(Markdown(resp.content))
    print("\n" + "="*80)

except Exception as e:
    print(f"Erreur Gemini: {e}")
    print("\nAssurez-vous que votre cle API est correcte.")
```

## Etape 8: Tests Avances - Sentiment Analysis

Testons l'analyse de sentiment avec notre LLM Gemini.

```python
# Definir un systeme de classification de sentiments
sentiment_system = """Vous etes un expert en analyse de sentiment.
Pour chaque texte donne, vous devez:
1. Determiner le sentiment (Positif, Negatif, Neutre)
2. Donner un score de confiance (0-100)
3. Justifier votre analyse en 1-2 phrases

Repondez au format:
Sentiment: [Positif/Negatif/Neutre]
Confiance: [score]%
Justification: [explication]"""

# Textes de test
test_texts = [
    "J'adore ce produit! C'est incroyable et je le recommande vivement a tous.",
    "Ce service est horrible. J'ai eu une mauvaise experience et je suis tres decu.",
    "Le produit a ete livre hier. Il fonctionne correctement."
]

print("="*80)
print("TEST: ANALYSE DE SENTIMENT AVEC GEMINI")
print("="*80)

try:
    for i, text in enumerate(test_texts, 1):
        print(f"\nTexte {i}: {text}")
        print("-" * 80)

        messages = [
            ("system", sentiment_system),
            ("human", text)
        ]

        resp = llm.invoke(messages)
        print("Reponse:")
        print(resp.content)

except Exception as e:
    print(f"Erreur: {e}")
```
