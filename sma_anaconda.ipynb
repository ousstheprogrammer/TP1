## 🔤 ÉTAPE 2 — Tokenisation avec tiktoken

La tokenisation est le processus qui découpe le texte en **tokens** (unités de base) avant de les envoyer au LLM.

````python
import tiktoken

# Charger le tokenizer associé à GPT-4o
tokenizer = tiktoken.encoding_for_model("gpt-4o")
print("Nom de l'encodage :", tokenizer.name)
````

````python
# Charger directement par nom d'encodage
tokenizer = tiktoken.get_encoding("o200k_base")
````

````python
# Définir un prompt
prompt = "Vous êtes un expert en marketing digital. C'est quoi un digital marketer ?"
print("Prompt :", prompt)
````

````python
# Encoder le prompt en tokens
tokens = tokenizer.encode(prompt)
print("IDs des tokens :", tokens)
````

````python
# Afficher le nombre de tokens
print("Nombre de tokens :", len(tokens))
````

````python
# Afficher chaque token individuellement
print("Découpage en tokens :")
for token in tokens:
    t = tokenizer.decode_single_token_bytes(token).decode("utf-8", errors="replace")
    print(t, end="|")
print()
````
